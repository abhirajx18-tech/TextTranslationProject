```python
import re, unicodedata, os
import pandas as pd
from datasets import load_dataset, Dataset, DatasetDict

ds = load_dataset("Helsinki-NLP/opus-100", "en-hi")

def to_df(split):
    return pd.DataFrame({
        "en": [x["en"] for x in split["translation"]],
        "hi": [x["hi"] for x in split["translation"]],
    })

def normalize(s):
    s = unicodedata.normalize("NFC", s)
    return re.sub(r"\s+", " ", s).strip()

DEV = re.compile(r"[\u0900-\u097F]")
LAT = re.compile(r"[A-Za-z]")

def script_frac(s, pattern):
    letters = [c for c in s if c.isalpha()]
    if not letters:
        return 0.0
    return sum(1 for c in letters if pattern.match(c)) / len(letters)

def clean(df):
    df = df.copy()
    df["en"] = df["en"].map(normalize)
    df["hi"] = df["hi"].map(normalize)
    df = df[(df.en != "") & (df.hi != "")]
    df = df[df.en.str.lower() != df.hi.str.lower()]          # copied text

    en_len = df.en.str.split().str.len()
    hi_len = df.hi.str.split().str.len()
    ratio = hi_len / en_len
    df = df[(en_len >= 2) & (en_len <= 50) & (hi_len >= 2) & (hi_len <= 60)]
    df = df[(ratio.loc[df.index] >= 0.4) & (ratio.loc[df.index] <= 2.5)]

    # Hindi side must be mostly Devanagari; English side mostly Latin
    df = df[df.hi.map(lambda s: script_frac(s, DEV)) >= 0.8]
    df = df[df.en.map(lambda s: script_frac(s, LAT)) >= 0.8]

    df = df.drop_duplicates(subset=["en", "hi"])
    df = df.drop_duplicates(subset=["en"])                    # one translation per source
    return df.reset_index(drop=True)

train_raw = to_df(ds["train"])
val_raw   = to_df(ds["validation"])
test_raw  = to_df(ds["test"])

train_c = clean(train_raw)
val_c   = clean(val_raw)
test_c  = clean(test_raw)

print(f"train: {len(train_raw)} -> {len(train_c)}")
print(f"val:   {len(val_raw)} -> {len(val_c)}")
print(f"test:  {len(test_raw)} -> {len(test_c)}")

# Remove any test/val sentences that leaked into train
held_out = set(val_c.en) | set(test_c.en) | set(test_raw.en) | set(val_raw.en)
train_c = train_c[~train_c.en.isin(held_out)].reset_index(drop=True)
print("train after leakage removal:", len(train_c))

# Subsample for training on 8 GB VRAM
train_sub = train_c.sample(n=min(100_000, len(train_c)), random_state=42).reset_index(drop=True)

# Save
os.makedirs("data/processed", exist_ok=True)
train_sub.to_csv("data/processed/train_en_hi.csv", index=False)
val_c.to_csv("data/processed/val_en_hi.csv", index=False)
test_c.to_csv("data/processed/test_clean_en_hi.csv", index=False)
test_raw.to_csv("data/processed/test_raw_en_hi.csv", index=False)

print(train_sub.sample(10, random_state=1).to_string())
```

```python
import re, os

# 1) English side must contain NO Devanagari (catches glued-together rows)
has_dev = train_c.en.str.contains(r"[\u0900-\u097F]", regex=True)

# 2) Religious-domain markers (English side)
REL_EN = re.compile(
    r"\b(Allah|thy|thee|thou|Muhammad|Messenger|disbelie\w*|unbelievers?|believers?|"
    r"Paradise|Hellfire|Satan|Pharaoh|Prophets?|revelations?|Resurrection|"
    r"transgressors?|wrongdoers?|Lord of|Gracious|Merciful|hath|verily)\b",
    re.IGNORECASE)
# 3) Religious-domain markers (Hindi side)
REL_HI = re.compile(
    r"(अल्लाह|ख़ुदा|खुदा|परवरदिगार|इबराहीम|फ़िरऔन|फिरौन|क़यामत|क़ियामत|जन्नत|जहन्नम|काफ़िर|रसूल)")

is_rel = train_c.en.str.contains(REL_EN) | train_c.hi.str.contains(REL_HI)

# 4) Brackets/quotes at the start often indicate Quran translations
is_bracketed = train_c.en.str.contains(r"\[[^\]]+\]") | train_c.en.str.startswith("(")

print("Devanagari in English column:", has_dev.sum())
print("Religious markers:           ", is_rel.sum())
print("Bracket style:               ", is_bracketed.sum())

train_c2 = train_c[~(has_dev | is_rel | is_bracketed)].reset_index(drop=True)
print("train after domain filtering:", len(train_c2))

train_sub = train_c2.sample(n=min(100_000, len(train_c2)), random_state=42).reset_index(drop=True)
train_sub.to_csv("data/processed/train_en_hi.csv", index=False)
print(train_sub.sample(15, random_state=3).to_string())
```

```python
import re

REL_EN2 = re.compile(
    r"\b(cometh|ye|unto|wherein|whosoever|shalt|doth|hast|wilt|punishment|penalty|"
    r"torment|grievous|usury|devour\w*|woe|guidance|guided|Lord|sinners?|righteous|"
    r"angels?|Hell|Judgment|those who (?:believe|disbelieve|reject)|disbelief|"
    r"Day of|forbidden|commandments?|worship\w*|God's)\b", re.IGNORECASE)

REL_HI2 = re.compile(r"(अज़ाब|अजाब|नेकी|ईमान|कुफ़्र|कुफ्र|मोमिन|नमाज़|ज़कात|जकात|क़ौम|मुत्तक़ी|तारीफ़|बदकार|गुनाहगार)")

UI_CHARS = re.compile(r"[%_&<>{}@#\\|=/~^]")
QUOTE_START = re.compile(r'^["“\']')

def domain_filter(df):
    en_words = df.en.str.split().str.len()
    hi_words = df.hi.str.split().str.len()
    bad = (
        df.en.str.contains(REL_EN2) | df.hi.str.contains(REL_HI2) |
        df.en.str.contains(UI_CHARS) | df.hi.str.contains(UI_CHARS) |
        df.en.str.contains(r"\[[^\]]+\]") | df.en.str.startswith("(") |
        df.en.str.contains(r"\.\.\.|\.\.") | df.hi.str.contains(r"\.\.\.|\.\.") |
        (en_words < 3) | (en_words > 25) | (hi_words > 35) |
        df.en.str.contains(r"\d{3,}")
    )
    return df[~bad].reset_index(drop=True), bad.sum()

train_c3, n_bad = domain_filter(train_c)
print("train removed in pass 2:", n_bad, "| remaining:", len(train_c3))

val_c3, _  = domain_filter(val_c)
test_c3, _ = domain_filter(test_c)
print("val:", len(val_c3), "| test:", len(test_c3))

train_sub = train_c3.sample(n=min(100_000, len(train_c3)), random_state=42).reset_index(drop=True)

train_sub.to_csv("data/processed/train_en_hi.csv", index=False)
val_c3.to_csv("data/processed/val_en_hi.csv", index=False)
test_c3.to_csv("data/processed/test_clean_en_hi.csv", index=False)

print(train_sub.sample(20, random_state=7).to_string())
```

```python
import pandas as pd, torch, sacrebleu
from transformers import MarianMTModel, MarianTokenizer

device = "cuda"
model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name).to(device).eval()

def translate_batch(sentences, batch_size=32, num_beams=4):
    outputs = []
    for i in range(0, len(sentences), batch_size):
        batch = sentences[i:i+batch_size]
        inputs = tokenizer(batch, return_tensors="pt", padding=True,
                           truncation=True, max_length=128).to(device)
        with torch.no_grad():
            gen = model.generate(**inputs, num_beams=num_beams, max_length=128)
        outputs += tokenizer.batch_decode(gen, skip_special_tokens=True)
    return outputs

def evaluate(df, name):
    hyps = translate_batch(list(df.en))
    refs = list(df.hi)
    bleu = sacrebleu.corpus_bleu(hyps, [refs], tokenize="none")
    chrf = sacrebleu.corpus_chrf(hyps, [refs])
    print(f"{name}: BLEU = {bleu.score:.2f} | chrF = {chrf.score:.2f}")
    return hyps

test_clean = pd.read_csv("data/processed/test_clean_en_hi.csv")
test_raw   = pd.read_csv("data/processed/test_raw_en_hi.csv")

hyp_clean = evaluate(test_clean, "Baseline, clean test")
hyp_raw   = evaluate(test_raw,   "Baseline, raw test")

# Save outputs for error analysis later
test_clean.assign(hyp=hyp_clean).to_csv("data/processed/baseline_clean_outputs.csv", index=False)
```

```python
SOFT = re.compile(
    r"\b(files?|folders?|directory|directories|windows?|servers?|users?|click|menu|"
    r"buttons?|dialog|default|settings?|configur\w*|install\w*|plugins?|cookies?|"
    r"desktop|disk|network|password|login|database|browser|application|program|"
    r"software|print\w*|selected|unable|failed|could not|cannot|error|version|"
    r"startup|panel|icon|toolbar|tooltip|screen|pixel|colou?rs?|Evolution|GNOME|KDE|Mozilla)\b",
    re.IGNORECASE)

def strict_filter(df):
    df, _ = domain_filter(df)                      # pass-2 filter (defined earlier)
    bad = (df.en.str.contains(REL_EN) | df.hi.str.contains(REL_HI) |
           df.en.str.contains(SOFT))
    return df[~bad].reset_index(drop=True)

train_c4 = strict_filter(train_c)
val_c4   = strict_filter(val_c)
test_c4  = strict_filter(test_c)
print("train:", len(train_c4), "| val:", len(val_c4), "| test:", len(test_c4))

train_sub = train_c4.sample(n=min(100_000, len(train_c4)), random_state=42).reset_index(drop=True)
train_sub.to_csv("data/processed/train_en_hi.csv", index=False)
val_c4.to_csv("data/processed/val_en_hi.csv", index=False)
test_c4.to_csv("data/processed/test_clean_en_hi.csv", index=False)

print(train_sub.sample(20, random_state=11).to_string())
```

```python
import gc, torch, pandas as pd
from datasets import Dataset
from transformers import (MarianMTModel, MarianTokenizer, DataCollatorForSeq2Seq,
                          Seq2SeqTrainingArguments, Seq2SeqTrainer)

# free GPU memory from the baseline model
for name in ["model", "trainer"]:
    if name in globals():
        del globals()[name]
gc.collect(); torch.cuda.empty_cache()

model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = MarianTokenizer.from_pretrained(model_name)
train_model = MarianMTModel.from_pretrained(model_name)

def load(path):
    df = pd.read_csv(path, keep_default_na=False)
    df = df[(df.en.str.strip() != "") & (df.hi.str.strip() != "")]
    return Dataset.from_pandas(df[["en", "hi"]].reset_index(drop=True))

train_ds = load("data/processed/train_en_hi.csv")
val_ds   = load("data/processed/val_en_hi.csv")

MAX_LEN = 96
def tokenize(batch):
    return tokenizer(batch["en"], text_target=batch["hi"],
                     max_length=MAX_LEN, truncation=True)

train_tok = train_ds.map(tokenize, batched=True, remove_columns=["en", "hi"])
val_tok   = val_ds.map(tokenize, batched=True, remove_columns=["en", "hi"])

collator = DataCollatorForSeq2Seq(tokenizer, model=train_model)

args = Seq2SeqTrainingArguments(
    output_dir="models/en-hi-ft",
    learning_rate=2e-5,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=64,
    num_train_epochs=3,
    weight_decay=0.01,
    warmup_steps=300,
    bf16=True,                       # more stable than fp16 for Marian
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    logging_steps=200,
    dataloader_num_workers=0,        # avoids hangs on Windows
    report_to="none",
)

trainer = Seq2SeqTrainer(
    model=train_model,
    args=args,
    train_dataset=train_tok,
    eval_dataset=val_tok,
    data_collator=collator,
    processing_class=tokenizer,
)

trainer.train()
trainer.save_model("models/en-hi-final")
tokenizer.save_pretrained("models/en-hi-final")
```

```python
import re, sacrebleu, pandas as pd

hand = pd.read_csv("data/processed/hand_results.csv", keep_default_na=False)

def norm(s):
    s = s.replace("।", ".").replace("॥", ".")
    s = re.sub(r"\s+([.?!,])", r"\1", s)     # remove space before punctuation
    s = re.sub(r"[.?!,]", "", s)             # drop punctuation entirely
    s = s.replace("़", "")                    # nukta variants (ज़/ज, फ़/फ)
    return re.sub(r"\s+", " ", s).strip()

refs = [norm(x) for x in hand.hi]
for name in ["base", "ft"]:
    hyps = [norm(x) for x in hand[name]]
    bleu = sacrebleu.corpus_bleu(hyps, [refs], tokenize="none")
    chrf = sacrebleu.corpus_chrf(hyps, [refs])
    print(f"{name:5s} normalized BLEU = {bleu.score:.2f} | chrF = {chrf.score:.2f}")

# per-sentence comparison
wins = {"base": 0, "ft": 0, "tie": 0}
for r, b, f in zip(refs, hand.base, hand.ft):
    sb = sacrebleu.sentence_chrf(norm(b), [r]).score
    sf = sacrebleu.sentence_chrf(norm(f), [r]).score
    wins["ft" if sf > sb else "base" if sb > sf else "tie"] += 1
print(wins)
```

```python
%%writefile app.py
import re, os, json
import requests
import torch
import gradio as gr
from transformers import MarianMTModel, MarianTokenizer
from wordfreq import zipf_frequency

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
LANGS = {"English": "en", "Hindi": "hi", "French": "fr"}

MODEL_PATHS = {
    ("en", "hi"): "models/en-hi-final",              # your fine-tuned model
    ("hi", "en"): "Helsinki-NLP/opus-mt-hi-en",
    ("en", "fr"): "Helsinki-NLP/opus-mt-en-fr",
    ("fr", "en"): "Helsinki-NLP/opus-mt-fr-en",
}

_cache = {}  # models are loaded lazily, then reused

def get_model(src, tgt):
    key = (src, tgt)
    if key not in _cache:
        path = MODEL_PATHS[key]
        tok = MarianTokenizer.from_pretrained(path)
        mdl = MarianMTModel.from_pretrained(path).to(DEVICE).eval()
        _cache[key] = (tok, mdl)
    return _cache[key]

def _translate_lines(lines, src, tgt):
    tok, mdl = get_model(src, tgt)
    batch = tok(lines, return_tensors="pt", padding=True,
                truncation=True, max_length=256).to(DEVICE)
    with torch.no_grad():
        out = mdl.generate(**batch, num_beams=4, max_length=256)
    return tok.batch_decode(out, skip_special_tokens=True)

# ================= name protection =================
# Placeholder names the model already translates well, with their Hindi forms
PLACEHOLDERS = [("Rahul", "राहुल"), ("Priya", "प्रिया"), ("Amit", "अमित"),
                ("Neha", "नेहा"), ("Rohan", "रोहन")]

STRONG_INTRO = re.compile(r"\b(?:my name is|name's|named|call me)\b[,:]?\s+([A-Za-z][A-Za-z'\-]*)", re.I)
WEAK_INTRO = re.compile(r"\b(?:i am|i'm|this is|meet)\b[,:]?\s+([A-Za-z][A-Za-z'\-]*)", re.I)
GREET = re.compile(r"\b(?:hi|hello|hey|dear|thanks|thank you)\b[,:]?\s+([A-Za-z][A-Za-z'\-]*)", re.I)
WORD = re.compile(r"[A-Za-z][A-Za-z'\-]*")
BRACKET = re.compile(r"\[([^\]]+)\]")

NAMES_FILE = "names.json"          # manual corrections (always win)
AUTO_FILE = "auto_names.json"      # cache of automatic results

def zipf(w):
    return zipf_frequency(w.lower(), "en")

def load_names():
    if os.path.exists(NAMES_FILE):
        with open(NAMES_FILE, encoding="utf-8") as f:
            return {k.lower(): v for k, v in json.load(f).items()}
    return {}

def _load_auto():
    if os.path.exists(AUTO_FILE):
        with open(AUTO_FILE, encoding="utf-8") as f:
            return json.load(f)
    return {}

def is_known(w):
    k = w.lower()
    return k in load_names() or k in _load_auto()

def _local_translit(word):
    """IndicXlit helper service (xlit_server.py). Returns None if it isn't running."""
    try:
        r = requests.get("http://127.0.0.1:5005/", params={"w": word}, timeout=2)
        cand = r.json().get("hi", "")
        if re.fullmatch(r"[\u0900-\u097F]+", cand):
            return cand
    except Exception:
        pass
    return None

def _online_translit(word):
    """Latin -> Devanagari via Google Input Tools (unofficial endpoint)."""
    try:
        r = requests.get(
            "https://inputtools.google.com/request",
            params={"text": word, "itc": "hi-t-i0-und", "num": 1,
                    "cp": 0, "cs": 1, "ie": "utf-8", "oe": "utf-8", "app": "demopage"},
            timeout=3)
        data = r.json()
        if data[0] == "SUCCESS":
            cand = data[1][0][1][0]
            if re.fullmatch(r"[\u0900-\u097F]+", cand):
                return cand
    except Exception:
        pass
    return None

def to_devanagari(name):
    key = name.lower()
    manual = load_names()
    if key in manual:                       # 1. manual override
        return manual[key]
    auto = _load_auto()
    if key in auto:                         # 2. cache
        return auto[key]
    result = _local_translit(name) or _online_translit(name)   # 3. IndicXlit, then online
    if result:
        auto[key] = result
        with open(AUTO_FILE, "w", encoding="utf-8") as f:
            json.dump(auto, f, ensure_ascii=False, indent=2)
        return result
    return name                             # 4. safe fallback: keep Latin spelling

def find_names(s):
    spans = []   # (start, end, name_text)
    for m in BRACKET.finditer(s):                       # explicit: [Abhiraj]
        spans.append((m.start(), m.end(), m.group(1)))
    for m in STRONG_INTRO.finditer(s):                  # "my name is X": always a name
        spans.append((m.start(1), m.end(1), m.group(1)))
    for pat in (WEAK_INTRO, GREET):                     # "I am X", "Hi X"
        for m in pat.finditer(s):
            w = m.group(1)
            if is_known(w) or zipf(w) < 3.2:
                spans.append((m.start(1), m.end(1), w))
    for m in WORD.finditer(s):                          # capitalised mid-sentence
        w = m.group(0)
        before = s[:m.start()].rstrip()
        if (m.start() > 0 and w[0].isupper() and w != "I" and "'" not in w
                and before and before[-1] not in ".!?\"“"
                and (is_known(w) or zipf(w) < 3.5)):
            spans.append((m.start(), m.end(), w))
    spans.sort()
    merged = []                                         # drop overlaps
    for sp in spans:
        if merged and sp[0] < merged[-1][1]:
            continue
        merged.append(sp)
    return merged

def mask(sentence, tgt):
    spans = find_names(sentence)[:len(PLACEHOLDERS)]
    if not spans:
        return sentence, []
    out, last, mapping = [], 0, []
    for i, (a, b, txt) in enumerate(spans):
        ph_en, ph_hi = PLACEHOLDERS[i]
        out.append(sentence[last:a]); out.append(ph_en); last = b
        if tgt == "hi":
            mapping.append((ph_hi, to_devanagari(txt)))
        else:
            mapping.append((ph_en, txt))
    out.append(sentence[last:])
    return "".join(out), mapping

def unmask(text, mapping):
    for expected, replacement in mapping:
        if expected not in text:
            return None          # placeholder was lost: caller falls back
        text = text.replace(expected, replacement)
    return text

def clean_english(s):
    """Fix common typing habits before translation."""
    s = s.strip()
    if not s:
        return s
    s = re.sub(r"\bi\b", "I", s)                                     # "i" -> "I"
    s = re.sub(r"\bi'm\b", "I'm", s, flags=re.I)
    s = re.sub(r"^(hi|hello|hey)\s+(?!,)", lambda m: m.group(1) + ", ", s, flags=re.I)
    s = s[0].upper() + s[1:]                                         # capital first letter
    if s[-1] not in ".!?":                                           # end punctuation
        s += "."
    return s

def translate_en_out(lines, tgt):
    """English -> hi/fr with input cleanup and name protection."""
    lines = [clean_english(l) for l in lines]
    masked = [mask(l, tgt) for l in lines]
    outs = _translate_lines([m[0] for m in masked], "en", tgt)
    final = []
    for line, o, (_, mp) in zip(lines, outs, masked):
        if not mp:
            final.append(o); continue
        r = unmask(o, mp)
        if r is None:
            plain = BRACKET.sub(r"\1", line)
            r = _translate_lines([plain], "en", tgt)[0]
        final.append(r)
    return final
# ===================================================

def translate_lines(lines, src, tgt):
    if src == "en" and tgt in ("hi", "fr"):
        return translate_en_out(lines, tgt)
    if (src, tgt) in MODEL_PATHS:
        return _translate_lines(lines, src, tgt)
    mid = _translate_lines(lines, src, "en")            # hi<->fr via English
    if tgt in ("hi", "fr"):
        return translate_en_out(mid, tgt)
    return _translate_lines(mid, "en", tgt)

def translate(text, src_name, tgt_name):
    if not text or not text.strip():
        return "Please enter some text."
    src, tgt = LANGS[src_name], LANGS[tgt_name]
    if src == tgt:
        return "Source and target languages are the same. Please choose different languages."
    try:
        lines = text.split("\n")
        idx = [i for i, l in enumerate(lines) if l.strip()]
        outs = translate_lines([lines[i].strip() for i in idx], src, tgt)
        result = list(lines)
        for i, o in zip(idx, outs):
            result[i] = o
        note = "  (via English)" if (src, tgt) not in MODEL_PATHS else ""
        return "\n".join(result) + (f"\n\n[{src_name} → {tgt_name}{note}]" if note else "")
    except Exception as e:
        return f"Translation failed: {e}"

with gr.Blocks(title="Text Translator") as demo:
    gr.Markdown("# Text Translator\nEnglish, Hindi and French. Pick the languages, type your text, and press Translate.\n\n"
                "Tip: put a name in [square brackets] to make sure it is kept as a name.")
    with gr.Row():
        src = gr.Dropdown(list(LANGS), value="English", label="Source language")
        swap = gr.Button("⇄ Swap", scale=0)
        tgt = gr.Dropdown(list(LANGS), value="Hindi", label="Target language")
    inp = gr.Textbox(lines=6, label="Text to translate", placeholder="Type or paste text here...")
    btn = gr.Button("Translate", variant="primary")
    out = gr.Textbox(lines=6, label="Translation", interactive=False)

    btn.click(translate, [inp, src, tgt], out)
    inp.submit(translate, [inp, src, tgt], out)
    swap.click(lambda s, t: (t, s), [src, tgt], [src, tgt])

if __name__ == "__main__":
    demo.launch()
```

```python
%%writefile xlit_server.py
import os
os.environ["TORCH_FORCE_NO_WEIGHTS_ONLY_LOAD"] = "1"   # must be set before importing torch/fairseq

import json, warnings
warnings.filterwarnings("ignore")
from http.server import BaseHTTPRequestHandler, HTTPServer
from urllib.parse import urlparse, parse_qs
from ai4bharat.transliteration import XlitEngine

engine = XlitEngine("hi", beam_width=10, src_script_type="en")
print("IndicXlit ready on http://127.0.0.1:5005")

class H(BaseHTTPRequestHandler):
    def do_GET(self):
        word = parse_qs(urlparse(self.path).query).get("w", [""])[0]
        try:
            res = engine.translit_word(word, topk=1)["hi"][0]
        except Exception:
            res = ""
        body = json.dumps({"hi": res}, ensure_ascii=False).encode("utf-8")
        self.send_response(200)
        self.send_header("Content-Type", "application/json; charset=utf-8")
        self.end_headers()
        self.wfile.write(body)
    def log_message(self, *a):
        pass

HTTPServer(("127.0.0.1", 5005), H).serve_forever()
```
